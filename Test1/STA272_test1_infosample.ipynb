## Sample Question 2: Linear Regression (5 points)

**Dataset:** grademployment.csv

**Task:** Using the graduate employment dataset, predict CGPA from other factors.

1. Load the dataset and select relevant variables (cgpa, internship, extracurriculars)
2. Split the data into train (70%) and test (30%) sets with `random_state=42`
3. Fit an OLS linear regression model predicting cgpa from internship and extracurriculars
4. Interpret the coefficient for internship

**Student Template (Fill in the `...`):**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import statsmodels.api as sm

# Load the dataset
grad_df = pd.read_csv('...')

# Select variables
X = grad_df[['...', '...']]
y = grad_df['...']

# Task 1: Train/test split (70/30)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=..., random_state=...)

print(f"Training set: {len(X_train)} observations")
print(f"Test set: {len(X_test)} observations")

# Task 2: Fit OLS model
X_train_const = sm.add_constant(...)
model = sm.OLS(..., ...).fit()

print("\n")
print(model.summary())

# Task 3: Get coefficient for internship
coef_internship = model.params['...']
print(f"\nCoefficient for internship: {coef_internship:.4f}")
```

### Solution Explanation

**Key Concepts:**

1. **Train/Test Split:**
   - `test_size=0.30` means 30% of data goes to test set, 70% to training
   - `random_state=42` ensures reproducible splits (same split every time)
   - Always specify random_state for consistency!

2. **Adding Intercept with `sm.add_constant()`:**
   - Statsmodels does NOT automatically add an intercept term
   - Must call `sm.add_constant(X_train)` to add a column of 1's
   - This allows the model to fit: $y=\beta_0 + \beta_1x_1+ \beta_2x_2 + \epsilon$

3. **OLS Argument Order:**
   - **CRITICAL:** `sm.OLS(y, X)` — response variable FIRST, predictors SECOND
   - This is opposite of sklearn's order!
   - Wrong order is a very common mistake

4. **Interpreting the Coefficient:**
   - Internship is a binary variable (0 or 1)
   - If the coefficient is 0.25: "Students with internships have, on average, a 0.25 higher GPA compared to students without internships, holding other factors constant"
   - For binary predictors, the coefficient is the difference in outcome between groups

**Common Mistakes:**
- Forgetting `sm.add_constant()` → model forced through origin
- Wrong argument order in `sm.OLS()` → strange results
- Not setting `random_state` → different results each time
- Misinterpreting binary variable coefficients as rates instead of differences

```python
# COMPLETE SOLUTION
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import statsmodels.api as sm

# Load the dataset
grad_df = pd.read_csv('grademployment.csv')

# Select variables
X = grad_df[['internship', 'extracurriculars']]
y = grad_df['cgpa']

# Task 1: Train/test split (70/30)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)

print(f"Training set: {len(X_train)} observations")
print(f"Test set: {len(X_test)} observations")

# Task 2: Fit OLS model
X_train_const = sm.add_constant(X_train)
model = sm.OLS(y_train, X_train_const).fit()

print("\n")
print(model.summary())

# Task 3: Get coefficient for internship
coef_internship = model.params['internship']
print(f"\nCoefficient for internship: {coef_internship:.4f}")
print(f"\nInterpretation: Students with internships have, on average, a {coef_internship:.4f} higher GPA")
print("compared to students without internships, holding extracurricular participation constant.")
```

**Task:** Predict whether a graduate will be employed within 6 months based on their academic and experience factors.

### Solution Explanation

**Key Concepts:**

1. **Stratified Train/Test Split:**
   - `stratify=y` ensures both train and test sets have similar proportions of employed/unemployed
   - Critical for binary outcomes to avoid imbalanced splits
   - Without stratification, one set might have very different employment rates

2. **Logistic Regression with `sm.Logit()`:**
   - Used for **binary outcomes** (0/1, True/False, Yes/No)
   - Like OLS, the syntax is `sm.Logit(y, X)` — response first!
   - Still need `sm.add_constant(X)` for intercept

3. **Odds Ratios:**
   - Odds Ratio = `exp(coefficient)`
   - **Interpretation for continuous variables (cgpa):** If OR = 2.5:
     - "Each 1-point increase in GPA multiplies the odds of employment by 2.5"
   - **Interpretation for binary variables (internship):** If OR = 3.0:
     - "Students with internships have 3 times the odds of employment compared to those without"
   - OR > 1 → positive relationship (higher X, higher odds)
   - OR < 1 → negative relationship (higher X, lower odds)
   - OR = 1 → no relationship

4. **Making Predictions:**
   - `.predict()` returns **probabilities** between 0 and 1
   - Convert to binary predictions using threshold (typically 0.5):
     - Probability ≥ 0.5 → predict 1 (employed)
     - Probability < 0.5 → predict 0 (not employed)

5. **Accuracy:**
   - Proportion of correct predictions: (TP + TN) / Total
   - Simple metric but may be misleading with imbalanced classes

**Connection to Test:**
- On the actual test, you'll also:
  - Calculate confusion matrix to see false positives/negatives
  - Calculate precision and recall
  - Fit decision trees and compare performance
  - Interpret which features are most important

**Common Mistakes:**
- Using `sm.OLS()` instead of `sm.Logit()` for binary outcomes
- Forgetting `stratify=y` in train_test_split
- Not adding constant to TEST set before predictions
- Confusing coefficient with odds ratio (need to exponentiate!)
- Interpreting odds ratios for binary variables incorrectly

```python
# COMPLETE SOLUTION
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import statsmodels.api as sm
from sklearn.metrics import accuracy_score, confusion_matrix

# Load the dataset
grad_df = pd.read_csv('grademployment.csv')

X = grad_df[['cgpa', 'internship', 'extracurriculars']]
y = grad_df['employed_6months']

# Task 1: Train/test split with stratification
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

print(f"Training set: {len(X_train)} observations")
print(f"Test set: {len(X_test)} observations")
print(f"Employment rate in train: {y_train.mean():.3f}")
print(f"Employment rate in test: {y_test.mean():.3f}")

# Task 2: Fit logistic regression
X_train_const = sm.add_constant(X_train)
logit_model = sm.Logit(y_train, X_train_const).fit()

# Task 3: Calculate odds ratios
odds_ratios = pd.DataFrame({
    'Coefficient': logit_model.params,
    'Odds Ratio': np.exp(logit_model.params)
})
print("\nOdds Ratios:")
print(odds_ratios)

# Task 4: Make predictions on test set
X_test_const = sm.add_constant(X_test)
y_pred_prob = logit_model.predict(X_test_const)
y_pred = (y_pred_prob >= 0.5).astype(int)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\nTest Set Accuracy: {accuracy:.4f}")

# Additional: Show confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)

print("\n--- Example Interpretations ---")
print(f"CGPA OR = {odds_ratios.loc['cgpa', 'Odds Ratio']:.3f}")
print("Interpretation: Each 1-point increase in GPA multiplies the odds of employment by this amount.")
print(f"\nInternship OR = {odds_ratios.loc['internship', 'Odds Ratio']:.3f}")
print("Interpretation: Students with internships have this many times the odds of employment compared to those without.")
```
