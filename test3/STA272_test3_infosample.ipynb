## Sample Question 2: K-Means Clustering

**Task:** Apply K-Means clustering to the standardized recipe features.

1. Fit K-Means with k=3 on the standardized features and compute the silhouette score
2. Compute silhouette scores for k = 2 to 8 to find the best k
3. Fit K-Means with k=4 and create a cluster profile table showing the mean of each **original (unscaled)** feature per cluster

**Student Template (Fill in the `...`):**

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Task 1: K-Means with k=3
km3 = KMeans(n_clusters=..., init='k-means++', n_init=10, random_state=42)  # 1 mark
labels3 = km3.fit_predict(...)  # 1 mark

sil3 = silhouette_score(..., ...)  # 1 mark
print(f"Silhouette score (k=3): {sil3:.4f}")

# Task 2: Silhouette scores for k=2 to 8
sil_scores = []
for k in range(2, 9):
    km = KMeans(n_clusters=..., init='k-means++', n_init=10, random_state=42)  # 1 mark
    labels = km.fit_predict(...)  # 1 mark
    sil = silhouette_score(..., ...)  # 1 mark
    sil_scores.append(sil)
    print(f"k={k}: silhouette = {sil:.4f}")

best_k = range(2, 9)[np.argmax(sil_scores)]
print(f"\nBest k by silhouette: {best_k}")

# Task 3: K-Means with k=4, cluster profile
km4 = KMeans(n_clusters=4, init='k-means++', n_init=10, random_state=42)
recipes['cluster'] = km4.fit_predict(X_scaled)

profile = recipes.groupby('...')[cluster_features].mean()  # 1 mark
print("\nCluster profile (mean of original features):")
print(profile.round(2))
```

### Solution Explanation

**Key Concepts:**

1. **KMeans Parameters:**
   - `n_clusters`: number of clusters (k)
   - `init='k-means++'`: smart initialization that speeds convergence
   - `n_init=10`: run the algorithm 10 times with different seeds, keep the best
   - `random_state=42`: reproducibility

2. **`fit_predict(X_scaled)`:** Fits the model and returns cluster labels in one step. Always pass the **standardized** features

3. **Silhouette Score:**
   - Measures how similar each point is to its own cluster vs. other clusters
   - Range: -1 to 1. Higher is better
   - Pass the **same features** used for fitting (standardized) and the cluster labels
   - `silhouette_score(X_scaled, labels)` — not `silhouette_score(labels, X_scaled)`!

4. **Cluster Profile Table:**
   - Use `groupby('cluster')[features].mean()` to get the mean of each feature per cluster
   - Use the **original (unscaled)** features for interpretability — standardized means are all near 0 and not useful for interpretation

**Common Mistakes:**
- Passing unstandardized features to `fit_predict` — always use `X_scaled`
- Wrong argument order in `silhouette_score`: it's `silhouette_score(X, labels)`, not `silhouette_score(labels, X)`
- Reporting cluster profiles on standardized features instead of original features
- Forgetting `random_state=42` — results will differ each run

```python
# COMPLETE SOLUTION
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Task 1: K-Means with k=3
km3 = KMeans(n_clusters=3, init='k-means++', n_init=10, random_state=42)
labels3 = km3.fit_predict(X_scaled)

sil3 = silhouette_score(X_scaled, labels3)
print(f"Silhouette score (k=3): {sil3:.4f}")

# Task 2: Silhouette scores for k=2 to 8
sil_scores = []
for k in range(2, 9):
    km = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    labels = km.fit_predict(X_scaled)
    sil = silhouette_score(X_scaled, labels)
    sil_scores.append(sil)
    print(f"k={k}: silhouette = {sil:.4f}")

best_k = range(2, 9)[np.argmax(sil_scores)]
print(f"\nBest k by silhouette: {best_k}")

# Task 3: K-Means with k=4, cluster profile
km4 = KMeans(n_clusters=4, init='k-means++', n_init=10, random_state=42)
recipes['cluster'] = km4.fit_predict(X_scaled)

profile = recipes.groupby('cluster')[cluster_features].mean()
print("\nCluster profile (mean of original features):")
print(profile.round(2))
```

## Sample Question 3: Hierarchical Clustering

3. Cut the dendrogram at k=4 using `fcluster` and compute the silhouette score

### Solution Explanation

**Key Concepts:**

1. **Ward's Linkage:**
   - `linkage(X_scaled, method='ward')` computes agglomerative clustering
   - Ward's method minimizes the total within-cluster variance at each merge step
   - Always pass **standardized** features

2. **Dendrogram:**
   - `truncate_mode='lastp', p=30` shows only the last 30 merges for readability
   - The y-axis shows merge distances — look for large gaps to determine natural cluster count
   - A large jump in distance between merges suggests cutting at that level

3. **fcluster:**
   - `fcluster(Z, t=4, criterion='maxclust')` cuts the dendrogram to produce exactly 4 clusters
   - Note: `fcluster` returns labels starting at 1 (not 0 like K-Means)

4. **Comparing to K-Means:**
   - Both methods can produce different cluster assignments for the same k
   - Compare silhouette scores to see which partitioning is more coherent
   - Hierarchical clustering provides a full hierarchy (dendrogram), while K-Means gives a single partition

**Common Mistakes:**
- Wrong linkage method: use `'ward'` not `'Ward'` or `'wards'`
- Forgetting `criterion='maxclust'` in `fcluster` — without it, `t` is interpreted as a distance threshold, not a cluster count
- Using unstandardized features with `linkage`

```python
# COMPLETE SOLUTION
from scipy.cluster.hierarchy import linkage, fcluster, dendrogram
import matplotlib.pyplot as plt

# Task 1: Ward's linkage
Z = linkage(X_scaled, method='ward')

# Task 2: Dendrogram
plt.figure(figsize=(14, 6))
dendrogram(Z, truncate_mode='lastp', p=30, leaf_rotation=90, leaf_font_size=10)
plt.title('Hierarchical Clustering Dendrogram (Ward, last 30 merges)')
plt.xlabel('Cluster size')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()

# Task 3: Cut at k=4, silhouette score
labels_h = fcluster(Z, t=4, criterion='maxclust')
sil_h = silhouette_score(X_scaled, labels_h)
print(f"Hierarchical clustering silhouette (k=4): {sil_h:.4f}")
```

## Sample Question 4: TF-IDF & Text Clustering

**Task:** Build a TF-IDF representation from recipe ingredients and cluster recipes by text.

1. Load `all_recipes.csv` and `cuisines.csv`, merge them on `url`, and drop rows with missing ingredients
2. Build a TF-IDF matrix from the `ingredients` column with English stop words and `max_features=2000`

4. Compute the cosine similarity matrix for the first 5 recipes
5. Run K-Means (k=5) on the TF-IDF matrix and report the silhouette score

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Task 1: Load and merge
all_recipes = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-09-16/all_recipes.csv')
cuisines = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-09-16/cuisines.csv')

recipes_df = all_recipes.merge(cuisines[['url', 'country']], on='...', how='inner')  # 1 mark
recipes_df = recipes_df.dropna(subset=['...'])  # 1 mark
print(f"Merged dataset shape: {recipes_df.shape}")

# Task 2: TF-IDF
tfidf = TfidfVectorizer(stop_words='...', max_features=...)  # 1 mark
tfidf_matrix = tfidf.fit_transform(recipes_df['...'])  # 1 mark
print(f"TF-IDF matrix shape: {tfidf_matrix.shape}")

# Task 3: Top 5 terms for first recipe
feature_names = tfidf.get_feature_names_out()
first_row = tfidf_matrix[0].toarray().flatten()
top5_idx = first_row.argsort()[...:]  # 1 mark (get last 5 = highest 5)
print(f"\nTop 5 TF-IDF terms for first recipe:")
for idx in reversed(top5_idx):
    print(f"  {feature_names[idx]}: {first_row[idx]:.4f}")

# Task 4: Cosine similarity for first 5 recipes
cos_sim = cosine_similarity(tfidf_matrix[...])  # 1 mark
print("\nCosine similarity (first 5 recipes):")
print(pd.DataFrame(cos_sim.round(3)))

# Task 5: K-Means on TF-IDF
km_text = KMeans(n_clusters=..., init='k-means++', n_init=10, random_state=42)  # 1 mark
text_labels = km_text.fit_predict(...)  # 1 mark
sil_text = silhouette_score(tfidf_matrix, text_labels)
print(f"\nSilhouette score (k=5 on TF-IDF): {sil_text:.4f}")
```

### Solution Explanation

**Key Concepts:**

1. **Merging Datasets:** `merge(..., on='url', how='inner')` joins the two datasets on the shared URL column. Inner join keeps only recipes present in both datasets

2. **TfidfVectorizer:**
   - `stop_words='english'` removes common English words (the, is, and, etc.)
   - `max_features=2000` keeps only the 2000 most frequent terms
   - `.fit_transform(text_column)` learns the vocabulary and transforms text to a sparse matrix
   - Each row is a recipe, each column is a term, values are TF-IDF weights

3. **TF-IDF Weighting:**
   - TF (term frequency): how often a term appears in a document
   - IDF (inverse document frequency): down-weights terms that appear in many documents
   - Common ingredients (salt, oil) get low weights; distinctive ingredients get high weights

4. **Cosine Similarity:**
   - Measures angle between document vectors, ignoring magnitude
   - Range: 0 (completely different) to 1 (identical)
   - Diagonal is always 1 (each document is identical to itself)
   - Matrix is symmetric: similarity(A, B) = similarity(B, A)
   - Slice the sparse matrix: `tfidf_matrix[:5]` gets the first 5 rows

5. **Text Clustering:**
   - K-Means works directly on sparse TF-IDF matrices
   - Silhouette scores on text data are often lower than on numeric data — this is normal

**Common Mistakes:**
- Forgetting `stop_words='english'` — common words dominate the representation
- Wrong column in `fit_transform`: must pass the text column (`ingredients`), not the whole DataFrame
- Using `tfidf_matrix` instead of `tfidf_matrix[:5]` for the first 5 recipes
- Forgetting to drop rows with missing ingredients before TF-IDF

```python
# COMPLETE SOLUTION
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Task 1: Load and merge
all_recipes = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-09-16/all_recipes.csv')
cuisines = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-09-16/cuisines.csv')

recipes_df = all_recipes.merge(cuisines[['url', 'country']], on='url', how='inner')
recipes_df = recipes_df.dropna(subset=['ingredients'])
print(f"Merged dataset shape: {recipes_df.shape}")

# Task 2: TF-IDF
tfidf = TfidfVectorizer(stop_words='english', max_features=2000)
tfidf_matrix = tfidf.fit_transform(recipes_df['ingredients'])
print(f"TF-IDF matrix shape: {tfidf_matrix.shape}")

# Task 3: Top 5 terms for first recipe
feature_names = tfidf.get_feature_names_out()
first_row = tfidf_matrix[0].toarray().flatten()
top5_idx = first_row.argsort()[-5:]
print(f"\nTop 5 TF-IDF terms for first recipe:")
for idx in reversed(top5_idx):
    print(f"  {feature_names[idx]}: {first_row[idx]:.4f}")

# Task 4: Cosine similarity for first 5 recipes
cos_sim = cosine_similarity(tfidf_matrix[:5])
print("\nCosine similarity (first 5 recipes):")
print(pd.DataFrame(cos_sim.round(3)))

# Task 5: K-Means on TF-IDF
km_text = KMeans(n_clusters=5, init='k-means++', n_init=10, random_state=42)
text_labels = km_text.fit_predict(tfidf_matrix)
sil_text = silhouette_score(tfidf_matrix, text_labels)
print(f"\nSilhouette score (k=5 on TF-IDF): {sil_text:.4f}")
```

## Sample Question 5: Interpretation & Synthesis (Qualitative)

Answer the following questions. No code is required.

**(a)** Why is standardization important before applying K-Means clustering? What would happen if you applied K-Means to the unstandardized recipe features (calories, fat, carbs, protein, prep_time, cook_time)?

**(b)** Compare `CountVectorizer` and `TfidfVectorizer` for computing cosine similarity between recipe ingredient lists. Which method produces a more informative representation for distinguishing between recipes? Explain why.

**(c)** Suppose you build a shallow autoencoder with architecture: Input (2000) → Dense(32, relu) → Dense(2000, sigmoid). Explain:
   - What the autoencoder is trained to do
   - Why clustering on the 32-dimensional bottleneck representation might produce different results than clustering on the raw 2000-dimensional TF-IDF vectors

### Sample Answers

---

**(a) Standardization and K-Means:**

K-Means uses Euclidean distance to assign points to clusters. Without standardization, features with larger numerical ranges dominate the distance calculation. For example, `calories` ranges from roughly 50 to 1,100, while `cook_time` ranges from 0 to about 60. A small difference in calories (say, 200 calories) would contribute far more to the Euclidean distance than a large difference in cook_time (say, 50 minutes), even though both differences may be equally meaningful.

The result: clusters would be determined almost entirely by calories (and possibly fat/carbs), effectively ignoring the time features. Standardization (mean 0, std 1) puts all features on a comparable scale so that each feature contributes equally to the distance computation.

---

**(b) CountVectorizer vs. TfidfVectorizer:**

**TfidfVectorizer** produces a more informative representation. `CountVectorizer` represents each recipe as a vector of raw term counts — common ingredients like "salt", "oil", and "water" that appear in many recipes get high counts in nearly every document, making all recipes look similar to each other.

`TfidfVectorizer` applies IDF (inverse document frequency) weighting, which down-weights terms that appear in many documents and up-weights terms that are distinctive to specific recipes. This means common ingredients contribute less to the representation while cuisine-specific ingredients (e.g., "soy sauce", "cumin", "coconut milk") get higher weights.

As a result, TF-IDF cosine similarities better distinguish between recipes because they are driven by the ingredients that make recipes unique rather than the ingredients they all share. CountVectorizer cosine similarities tend to be higher and more uniform, providing less discrimination.

---

**(c) Autoencoder and bottleneck clustering:**

**What it learns:** The autoencoder is trained to **reconstruct** its input. It takes a 2000-dimensional TF-IDF vector, compresses it to 32 dimensions through the bottleneck layer, then tries to recreate the original 2000-dimensional vector. The loss function (e.g., binary crossentropy or MSE) penalizes reconstruction error, forcing the 32-dimensional bottleneck to capture the most important patterns in the data.

**Why clustering differs:** The 32-dimensional bottleneck is a learned, compressed representation that captures the most salient patterns needed for reconstruction. It may group related terms together and discard noise, producing a denser, more structured feature space. K-Means on this lower-dimensional space may find more meaningful clusters because:
- The curse of dimensionality is reduced (Euclidean distance is more meaningful in 32 dims than in 2000 dims)
- The autoencoder has learned to represent co-occurring ingredients as similar patterns
- Noisy or rare terms that add little information are effectively filtered out

However, the autoencoder's bottleneck representation is optimized for reconstruction, not for clustering specifically, so it may not always produce better clusters.
