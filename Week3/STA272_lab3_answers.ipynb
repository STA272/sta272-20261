## Linear Regression for Prediction

In this lab, you will practice building and evaluating a linear regression model for prediction using the movie dataset from lecture.

### Task #1

Read `movies_modeling.csv` into a `pandas` dataframe called `movies_df`.

The code cell below contains starter code. Fill in the ellipsis `(...)` with the required code.

```python
# fill in your answer for Task #1 in this cell

import pandas as pd
import numpy as np

movies_df = pd.read_csv('movies_modeling.csv')

#check your work
movies_df.head()
```

### Task #2

Create a log-transformed version of the `revenue` column and store it in a new column called `log_revenue` in `movies_df`. Use `np.log()` for the transformation.

```python
# fill in your answer for Task #2 in this cell

movies_df['log_revenue'] = np.log(movies_df['revenue'])
```

### Task #3

Create two new columns in `movies_df`:

1. A binary indicator variable called `comedy` that equals 1 if the movie's `genres` column contains the string `'Comedy'`, and 0 otherwise.

2. A binary indicator variable called `adventure` that equals 1 if the movie's `genres` column contains the string `'Adventure'`, and 0 otherwise.

Hint: use `.str.contains()` method on the genres column and `.astype(int)` to convert boolean to integer (0/1).

```python
# fill in your answer for Task #3 in this cell

movies_df['comedy'] = movies_df['genres'].fillna('').str.contains('Comedy').astype(int)
movies_df['adventure'] = movies_df['genres'].fillna('').str.contains('Adventure').astype(int)
```

### Task #4a

For our model, we will use `log_revenue` as the response (y) and `adventure`, `comedy`, `popularity`, `actor_avg_pop`, and `runtime` as predictors (X).

First, define a list called `key_vars` containing these variable names: `['log_revenue', 'adventure', 'comedy', 'popularity', 'actor_avg_pop', 'runtime']`.

Check the number of missing values in these key variables using `.isnull().sum()`. Then, create a new DataFrame called `movies_df_clean` by removing rows with missing values in these key variables using `dropna(subset=key_vars)`.

```python
# fill in your answer for Task #4a in this cell

key_vars = ['log_revenue', 'adventure', 'comedy', 'popularity', 'actor_avg_pop', 'runtime']
print("Missing values in key variables:")
print(movies_df[key_vars].isnull().sum())
```

```python
# fill in your answer for Task #4a here

movies_df_clean = movies_df.dropna(subset=key_vars)
```

### Task #4b

Using `sklearn.model_selection.train_test_split`, split the data into training and test sets with a **70-30 split** (i.e., `test_size=0.3`). Set your **seed** by specifying `random_state=42`.

Create the following variables:
- `X`: DataFrame containing the predictor columns `['adventure', 'comedy', 'popularity', 'actor_avg_pop', 'runtime']` from `movies_df_clean`
- `y`: Series containing the response column `'log_revenue'` from `movies_df_clean`
- `X_train`, `X_test`, `y_train`, `y_test`: The train/test splits returned by `train_test_split`

```python
# fill in your answer for Task #4b in this cell

from sklearn.model_selection import train_test_split

X = movies_df_clean[['adventure', 'comedy', 'popularity', 'actor_avg_pop', 'runtime']]
y = movies_df_clean['log_revenue']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

### Task #5a

Fit a linear regression model using `statsmodels` on the training data and store the fitted model in a variable called `model`. Print the model summary.

Hint: remember to add a constant term (intercept) using `sm.add_constant()`.

```python
# fill in your answer for Task#5a in this cell

import statsmodels.api as sm

X_train_const = sm.add_constant(X_train)
model = sm.OLS(y_train, X_train_const).fit()

print(model.summary())
```

### Task #5b 

Use `r2_score` to evaluate your model's test performance.

Create the following variables:
- `X_test_const`: The test predictors with a constant added using `sm.add_constant(X_test)`
- `y_pred`: Predicted values on the test set
- `r2`: The test $R^2$ computed with `r2_score`

Hints:
- `model.predict(...)` needs the test predictors with a constant.
- `r2_score` expects the true test values and your predictions.


```python
# fill in your answer for Task#5b in this cell

from sklearn.metrics import r2_score

# Predict on test set
X_test_const = sm.add_constant(X_test)
y_pred = model.predict(X_test_const)

r2 = r2_score(y_test, y_pred)

print(r2)
```
