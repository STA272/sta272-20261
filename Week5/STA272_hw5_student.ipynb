## Question #8: Understanding Tree Parameters and Overfitting

A decision tree learns rules by repeatedly splitting the data into smaller groups to make predictions. The settings you choose for the tree (called hyperparameters) control how complex the tree is allowed to become.

`max_depth`: the maximum number of split levels from the root to a leaf. Bigger `max_depth` results in a more complex tree.

`min_samples_leaf`: the minimum number of training examples that must be in a leaf node. Bigger values force leaves to be “supported” by more data, making the tree simpler and more stable.

Overfitting happens when a model fits the training data extremely well by learning noise or tiny quirks, instead of learning general patterns. An overfit model usually has:

- very high training performance (e.g., accuracy), but
- worse performance on new (unseen) data, because the rules it learned don’t generalize.

In decision trees, overfitting often occurs when the tree is allowed to become too deep or create very small leaves, leading to overly specific rules.
