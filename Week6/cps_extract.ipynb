# CPS Employment Data (Extract 3)

Individual-level microdata from [IPUMS CPS](https://cps.ipums.org/cps/) with expanded variables including race, citizenship, work status, firm size, income, and health insurance.

| Variable | Description | Values |
|----------|-------------|--------|
| `YEAR` | Survey year | 2020-2025 |
| `MONTH` | Survey month | 1-12 |
| `STATEFIP` | State (FIPS code) | 1-56 = states, 61-90 = grouped states, 99 = not identified |
| `METRO` | Metropolitan area status | 0=Not identified, 1=Non-metro, 2=Central city, 3=Not central city, 4=Central city unknown, 9=Missing |
| `FAMINC` | Family income | 100-150=Under $5k, 200-350=$5k-$10k, 400-490=$10k-$15k, 500-560=$15k-$20k, 600=$20k-$25k, 710=$25k-$30k, 720=$30k-$35k, 730=$35k-$40k, 740=$40k-$50k, 820=$50k-$60k, 830=$60k-$75k, 841=$75k-$100k, 842=$100k-$150k, 843=$150k+; 995-999=Missing |
| `AGE` | Age | 0-99 |
| `SEX` | Sex | 1=Male, 2=Female, 9=NIU |
| `RACE` | Race | 100=White, 200=Black, 300=AIAN, 651=Asian, 652=NHPI, 700=Other, 801+=Multiracial |
| `MARST` | Marital status | 1=Married spouse present, 2=Married spouse absent, 3=Separated, 4=Divorced, 5=Widowed, 6=Never married, 9=NIU |
| `NCHILD` | Number of own children in household | 0-9 |
| `CITIZEN` | Citizenship status | 1=Born in US, 2=Born in US outlying, 3=Born abroad of American parents, 4=Naturalized, 5=Not a citizen, 9=NIU |
| `NATIVITY` | Foreign birthplace or parentage | 0=Unknown, 1=Both parents native, 2=Father foreign, 3=Mother foreign, 4=Both parents foreign, 5=Foreign born |
| `EMPSTAT` | Employment status | 0=NIU, 1=Armed Forces, 10=At work, 12=Has job not at work, 20-22=Unemployed, 30-36=NILF |
| `OCC` | Occupation | 4-digit Census codes |
| `IND` | Industry | 4-digit Census codes |
| `CLASSWKR` | Class of worker | 13=Self-emp not inc, 14=Self-emp inc, 22=Private for-profit, 23=Private nonprofit, 25=Federal, 27=State, 28=Local, 29=Unpaid family |
| `WKSTAT` | Full/part time status | 10-15=Full-time, 20-22=Part-time economic, 40-42=Part-time non-economic, 50=Unemp FT, 60=Unemp PT, 99=NIU |
| `EDUC` | Educational attainment | 0-1=NIU, 73=HS diploma, 81=Some college, 91-92=Associate's, 111=Bachelor's, 123=Master's, 124=Professional, 125=Doctorate, 999=Missing |
| `FIRMSIZE` | Number of employees | 0=NIU, 1=Under 10, 2=10-24, 3=Under 25, 4=10-49, 5=25-99, 6=50-99, 7=100-499, 8=500-999, 9=1000+ |
| `INCWAGE` | Wage and salary income | Dollar amount; 99999999=NIU, 99999998=Missing |
| `ANYCOVLY` | Any health insurance last year | 1=No, 2=Yes, 99=NIU |

The data is in fixed-width format (`cps_00003.dat`), read using column positions from the IPUMS codebook.

## Read the fixed-width data file

```python
import pandas as pd
import numpy as np

# Column positions from the IPUMS SAS codebook (1-indexed, inclusive)
# Convert to 0-indexed half-open intervals for pandas
colspecs = [
    (0, 4),       # YEAR        1-4
    (4, 9),       # SERIAL      5-9
    (9, 11),      # MONTH       10-11
    (11, 25),     # CPSID       12-25
    (25, 26),     # ASECFLAG    26-26
    (26, 37),     # ASECWTH     27-37  (4 implied decimals)
    (37, 39),     # STATEFIP    38-39
    (39, 40),     # METRO       40-40
    (40, 43),     # FAMINC      41-43
    (43, 45),     # PERNUM      44-45
    (45, 59),     # CPSIDP      46-59
    (59, 74),     # CPSIDV      60-74
    (74, 85),     # ASECWT      75-85  (4 implied decimals)
    (85, 87),     # AGE         86-87
    (87, 88),     # SEX         88-88
    (88, 91),     # RACE        89-91
    (91, 92),     # MARST       92-92
    (92, 93),     # NCHILD      93-93
    (93, 94),     # CITIZEN     94-94
    (94, 95),     # NATIVITY    95-95
    (95, 97),     # EMPSTAT     96-97
    (97, 101),    # OCC         98-101
    (101, 105),   # IND         102-105
    (105, 107),   # CLASSWKR    106-107
    (107, 109),   # WKSTAT      108-109
    (109, 112),   # EDUC        110-112
    (112, 113),   # FIRMSIZE    113-113
    (113, 122),   # INCTOT      114-122
    (122, 130),   # INCWAGE     123-130
    (130, 132),   # ANYCOVLY    131-132
]

col_names = ['YEAR', 'SERIAL', 'MONTH', 'CPSID', 'ASECFLAG',
             'ASECWTH', 'STATEFIP', 'METRO', 'FAMINC', 'PERNUM',
             'CPSIDP', 'CPSIDV', 'ASECWT', 'AGE', 'SEX',
             'RACE', 'MARST', 'NCHILD', 'CITIZEN', 'NATIVITY',
             'EMPSTAT', 'OCC', 'IND', 'CLASSWKR', 'WKSTAT',
             'EDUC', 'FIRMSIZE', 'INCTOT', 'INCWAGE', 'ANYCOVLY']

cps = pd.read_fwf('cps_00003.dat', colspecs=colspecs, names=col_names, header=None)

# Adjust implied decimal columns (4 implied decimal places)
for col in ['ASECWTH', 'ASECWT']:
    cps[col] = cps[col] / 10000

print(f"Dataset shape: {cps.shape}")
cps.head()
```

## Clean the data

### Step 1: Remove NIU (Not In Universe) and missing/invalid codes

```python
# NIU and missing codes for each variable (from the IPUMS codebook)
niu_codes = {
    'EMPSTAT':  [0],                    # 0 = NIU
    'EDUC':     [0, 1, 999],            # 0 = NIU/no schooling, 1 = NIU/blank, 999 = Missing
    'SEX':      [9],                    # 9 = NIU
    'MARST':    [9],                    # 9 = NIU
    'RACE':     [999],                  # 999 = Blank
    'METRO':    [0, 9],                 # 0 = Not identified, 9 = Missing
    'FAMINC':   [995, 996, 997, 999],   # Missing / Refused / Don't know / Blank
    'CITIZEN':  [9],                    # 9 = NIU
    'NATIVITY': [0],                    # 0 = Unknown
    'WKSTAT':   [99],                   # 99 = NIU/NILF
    'ANYCOVLY': [99],                   # 99 = NIU
}

print(f"Before removing NIU/missing: {len(cps):,} rows")

cps_clean = cps.copy()

# Remove NIU/missing wage income values
cps_clean = cps_clean[~cps_clean['INCWAGE'].isin([99999999, 99999998])]

for col, codes in niu_codes.items():
    before = len(cps_clean)
    cps_clean = cps_clean[~cps_clean[col].isin(codes)]
    dropped = before - len(cps_clean)
    if dropped > 0:
        print(f"  Dropped {dropped:>7,} rows with NIU/missing in {col}")

print(f"After removing NIU/missing:  {len(cps_clean):,} rows")
```

### Step 2: Deduplicate

CPS uses a rotating panel design, so the same person can appear in multiple months. We keep only the first observation per person.

```python
print(f"Before deduplication: {len(cps_clean):,} rows")
cps_clean = cps_clean.drop_duplicates(subset='CPSIDP', keep='first')
print(f"After deduplication:  {len(cps_clean):,} rows")

print(f"\nEMPSTAT distribution:")
out = cps_clean['EMPSTAT'].value_counts().to_frame('count')
out['label'] = out.index.map(empstat_labels)
out['percent'] = (out['count'] / out['count'].sum() * 100).round(1)
print(out)
```

### Step 3: Recode predictors and target

```python
# Recode education into broader categories
def recode_educ(x):
    if x <= 71:              # less than high school
        return 1
    elif x == 73:            # high school diploma
        return 2
    elif x in [81, 91, 92]:  # some college or associate's
        return 3
    elif x == 111:           # bachelor's degree
        return 4
    elif x in [123, 124, 125]:  # graduate degree
        return 5
    else:
        return np.nan

cps_clean['educ_cat'] = cps_clean['EDUC'].apply(recode_educ)

educ_cat_labels = {1: 'Less than HS', 2: 'HS diploma', 3: 'Some college/Associates',
                   4: "Bachelor's", 5: 'Graduate'}

out = cps_clean['educ_cat'].value_counts().sort_index().to_frame('count')
out['label'] = out.index.map(educ_cat_labels)
out['percent'] = (out['count'] / out['count'].sum() * 100).round(1)
print(out)
```

```python
# Recode family income: Low (<$40k), Middle ($40k-$100k), High ($100k+)
# FAMINC codes: <=730 = under $40k, 740-841 = $40k-$100k, 842+ = $100k+
def recode_faminc(x):
    if x <= 730:      # under $40,000
        return 1      # Low
    elif x <= 841:    # $40,000 - $99,999
        return 2      # Middle
    else:             # $100,000+
        return 3      # High

cps_clean['faminc_cat'] = cps_clean['FAMINC'].apply(recode_faminc)

faminc_cat_labels = {1: 'Low (<$40k)', 2: 'Middle ($40k-$100k)', 3: 'High ($100k+)'}

out = cps_clean['faminc_cat'].value_counts().sort_index().to_frame('count')
out['label'] = out.index.map(faminc_cat_labels)
out['percent'] = (out['count'] / out['count'].sum() * 100).round(1)
print(out)
```

```python
# Recode metro: 1 = non-metro -> 0; 2,3,4 = metro -> 1
cps_clean['metro_binary'] = cps_clean['METRO'].map({1: 0, 2: 1, 3: 1, 4: 1})

# Recode sex: 1=Male -> 0, 2=Female -> 1
cps_clean['female'] = (cps_clean['SEX'] == 2).astype(int)

# Recode marital status: 1,2 = married; 3-7 = not married
cps_clean['married'] = (cps_clean['MARST'].isin([1, 2])).astype(int)

# Recode race into broader categories
def recode_race(x):
    if x == 100:
        return 'White'
    elif x == 200:
        return 'Black'
    elif x == 651:
        return 'Asian'
    else:
        return 'Other'

cps_clean['race_cat'] = cps_clean['RACE'].apply(recode_race)

# Recode citizenship: 1 = US citizen (born or naturalized), 0 = not a citizen
cps_clean['us_citizen'] = (cps_clean['CITIZEN'].isin([1, 2, 3, 4])).astype(int)

# Has children: 0 = no, 1 = yes
cps_clean['has_children'] = (cps_clean['NCHILD'] > 0).astype(int)

# Health insurance: 1 = no -> 0, 2 = yes -> 1
cps_clean['insured'] = (cps_clean['ANYCOVLY'] == 2).astype(int)

print(f"Female:       {cps_clean['female'].mean():.3f}")
print(f"Married:      {cps_clean['married'].mean():.3f}")
print(f"Metro:        {cps_clean['metro_binary'].mean():.3f}")
print(f"US citizen:   {cps_clean['us_citizen'].mean():.3f}")
print(f"Has children: {cps_clean['has_children'].mean():.3f}")
print(f"Insured:      {cps_clean['insured'].mean():.3f}")
print(f"\nRace distribution:")
print(cps_clean['race_cat'].value_counts())
```

```python
# Recode class of worker (target variable)
# Combines CLASSWKR, WKSTAT, and EMPSTAT to create categories
# that include part-time workers and unemployed

def recode_worker(row):
    empstat = row['EMPSTAT']
    classwkr = row['CLASSWKR']
    wkstat = row['WKSTAT']
    
    # Unemployed (EMPSTAT 20, 21, 22)
    if empstat in [20, 21, 22]:
        return 'Unemployed'
    
    # Part-time workers (WKSTAT 20-22 = PT economic, 40-42 = PT non-economic)
    if wkstat in [20, 21, 22, 40, 41, 42]:
        return 'Part-time'
    
    # Full-time employed by class of worker
    if classwkr in [13, 14]:
        return 'Self-employed'
    elif classwkr in [21, 22]:
        return 'Private for-profit'
    elif classwkr == 23:
        return 'Private nonprofit'
    elif classwkr in [25, 26, 27, 28]:
        return 'Government'
    else:
        return np.nan

cps_clean['worker_cat'] = cps_clean.apply(recode_worker, axis=1)

out = cps_clean['worker_cat'].value_counts().to_frame('count')
out['percent'] = (out['count'] / out['count'].sum() * 100).round(1)
print(out)
```

### Step 4: Select features and check for remaining missing values

```python
feature_cols = ['AGE', 'female', 'married', 'educ_cat', 'faminc_cat',
                'metro_binary', 'race_cat', 'us_citizen', 'has_children',
                'insured', 'OCC', 'IND',
                'INCWAGE', 'FIRMSIZE', 'NCHILD', 'YEAR', 'STATEFIP']
target_col = 'worker_cat'

cps_model = cps_clean[feature_cols + [target_col]].copy()

# Check for any remaining missing values
print("Missing values per column:")
print(cps_model.isnull().sum())
print(f"\nRows before dropping NaN: {len(cps_model):,}")

cps_model = cps_model.dropna()
print(f"Rows after dropping NaN:  {len(cps_model):,}")
print(f"\nTarget distribution:")
print(cps_model[target_col].value_counts())
cps_model.head(10)
```

## Recent graduates subset

Filter to ages 22-30 with a bachelor's degree or higher (educ_cat 4 or 5) for studying employment outcomes among recent university graduates.
