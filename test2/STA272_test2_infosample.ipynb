## Topics Covered in Weeks 6 and 7

### Week 6: Bagging and Random Forests

**Decision Tree Issues:**
- Overfitting in fully grown trees
- Difference between training and test accuracy
- Pre-pruning (stopping rules) vs. post-pruning
- Cost-complexity pruning using `ccp_alpha` parameter
- Cross-validation for selecting optimal tree size

**Ensemble Methods:**
- **Bootstrap aggregating (bagging):** averaging predictions from multiple trees trained on bootstrap samples
- **Random forests:** extension of bagging with random feature selection at each split
- Key parameters: `n_estimators` (number of trees), `max_features` (features per split)
- Why ensemble methods reduce variance while maintaining low bias

**Variable Importance:**
- Computing and interpreting variable importance based on mean decrease in Gini impurity (gain)
- Understanding which features contribute most to predictions

**Applications:**
- Binary classification (employment status)
- Regression (wage prediction with log transformation)
- Using `RandomForestClassifier` and `RandomForestRegressor`
- Evaluating models: accuracy, RMSE, R², confusion matrices
- Handling class imbalance with `class_weight='balanced'`

### Week 7: Introduction to Neural Networks

**Neural Network Fundamentals:**
- Neural networks as a general framework for prediction tasks
- Feedforward/multilayer perceptron (MLP) architecture
- Understanding that a neural network without hidden layers = linear/logistic regression
- How hidden layers enable learning nonlinear patterns
- Dense layers, units, activation functions (ReLU)

**Implementation in Python:**
- Using `MLPRegressor` for continuous outcomes
- Using `MLPClassifier` for binary and multiclass classification
- Feature scaling with `StandardScaler` (critical for neural networks)
- Key parameters: `hidden_layer_sizes`, `activation`, `solver`, `max_iter`
- Early stopping to prevent overfitting

**Applications and Evaluation:**
- CPS wage data (tabular prediction)
- MNIST handwritten digits (image classification)
- Converting images to numerical data (pixel values)
- Supervised learning workflow: train/test split, fit, evaluate
- Metrics: accuracy, confusion matrix, precision, recall, F1 score, RMSE, R²

**Ethical Considerations:**
- Fairness and bias in supervised machine learning models
- Subgroup evaluation (checking model performance across demographic groups)
- Potential for discrimination and amplifying historical bias
- Unequal error rates across protected groups
- Real-world implications of using models for decision-making

**Practical Insights:**
- Neural networks excel with complex, nonlinear patterns (e.g., images)
- For simple tabular data, improvement over logistic regression may be modest
- Infrastructure considerations: AI data centers, compute requirements, energy costs
