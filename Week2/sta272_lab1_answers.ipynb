## STA272 Lab #1

<https://en.wikipedia.org/wiki/List_of_most_popular_social_platforms> has a list of the most popular social platforms (at least 100 million users).  The table has been scraped and stored in `expected_social100.csv`.


### Task #1

Read `expected_social100.csv` into a `pandas` dataframe called `social100_df`.

```python
# answer
import pandas as pd


social100_df = pd.read_csv("expected_social100.csv")

social100_df.head()
```

### Task #2   
Clean the column labelled Monthly active users.  In `social100_df` store the values as a `float` in units of billions. For example, Douyin has 755 million users which should be stored as 0.755. Store this column in a variable called `active_users_float`.

```python
# answer

active_users_float = (social100_df["Monthly active users"].
    astype(str).
    str.extract(r'([\d.]+)', expand=False).
    astype(float))
```

### Task #3

The units of this column are either in millions of billions. Create a boolean variable called `billion` that is `True` if the users are recorded in billions or `False` if recorded in millions.

(Hint: use the values of `active_users_float`).

```python
# answer

billion = active_users_float < 4

billion
```

### Task #4

Some values of `active_users` are in billions and some values are in millions.  Create a new variable `active_users_units` whose values are `active_users` in billions.  This can be done by dividing values that are in the millions in `active_users` by 1000, since 1000 x 1 million = 1 billion. 

```python
# answer

active_users_units = active_users_float.where(billion, active_users_float / 1000)
```

### Task #5

Create a new dataframe called `social100_df2` that includes columns: `active_users_float`, `billion`, and `active_users_units`.

```python
social100_df2 = social100_df.copy()

social100_df2['active_users_float'] = active_users_float

social100_df2['billion'] = billion

social100_df2['active_users_units'] = active_users_units
```
